４. 在數學上0.1+0.2=0.3，可是下面的Python程式的答案好像不是0.3。怎麼解決？
```python
print(0.1+0.2)
```

程式：

Python
print(0.1+0.2)

輸出：

Python
0.30000000000000004
 

原因是：

Python
0.1

和

Python
0.2

都無法被二進位精確表示，因此相加後產生微小誤差。

解決方法：

Python
round(0.1+0.2, 1)

或

Python
import math
math.isclose(0.1+0.2, 0.3)

若是金融計算：

Python
from decimal import Decimal

使用 Decimal 較適合。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

科學計算：

使用 float 搭配 math.isclose() 是常見且合適的方法，因為運算速度快。

金融計算：

不建議使用 float，因為誤差可能累積造成金額錯誤。

較好的做法是：

Python
Decimal

或以整數儲存最小貨幣單位（如分、角）。

結論：

科學計算適合使用 float，金融計算應使用 Decimal 或整數表示金額。
