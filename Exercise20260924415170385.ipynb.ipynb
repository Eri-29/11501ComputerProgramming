## 第五部分：Python 整數與浮點數

執行下列程式：

```python
print(10**100)
print(10**1000)
```

回答：

1. 為什麼 Python 能表示如此大的整數？因為 int 採用任意精度（Arbitrary Precision）設計，可依需求增加儲存空間。
2. Python 的 `int` 是否具有固定的位元數限制？沒有。
3. 什麼是 Arbitrary Precision（任意精度）？指整數位數可隨數值大小動態擴充，而非固定長度。
4. CPython 使用什麼物件來實作 `int`？PyLongObject
5. Python 整數大小真正的限制是什麼？電腦的記憶體（RAM）與系統資源。

---

再執行：

```python
0.1 + 0.2
```

```python
0.1 + 0.2 == 0.3
```

```python
format(0.1, ".20f")
```

回答：

1. 為什麼結果與數學上的答案不同？因為 float 使用二進位浮點數儲存，而 0.1、0.2 無法被二進位精確表示，所以會產生微小誤差。
2. 浮點數是否能精確表示所有小數？不能。
3. 什麼是浮點數的精度（precision）？是指：浮點數能正確表示的有效數字位數。
4. 什麼是浮點數的範圍（range）？是指：浮點數能表示的最大值與最小值之間的區間。
5. 精度與範圍有何不同？精度（Precision）：能表示得多準確。
範圍（Range）：能表示多大或多小。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？浮點數（Floating Point）
2. 什麼是 FP64？64 位元雙精度浮點數。精度最高，常用於科學計算。
3. 什麼是 FP32？佔 32 bits（4 bytes），長期為深度學習標準格式。
4. 什麼是 FP16？佔 16 bits（2 bytes）記憶體需求為 FP32 的一半，運算速度較快
5. 什麼是 BF16？佔 16 bits（2 bytes）
與 FP32 有相近的數值範圍，常用於大型 AI 模型訓練佔 16 bits（2 bytes）
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？1.權重通常是小數，不是整數。2.神經網路需要大量浮點運算。3.int 佔用更多儲存空間且運算效率較差。

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？沒有固定位元數限制，可依需要擴充大小。
2. 它是否代表可以儲存無限大的整數？不是，仍受電腦記憶體限制。
3. CPython 如何實現這項特性？使用 PyLongObject，並透過動態配置記憶體儲存大整數。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？因為 int 採用任意精度（Arbitrary Precision）設計，可動態增加記憶體空間。
2. 為什麼 `float` 卻存在範圍限制？因為 float 使用固定 64 位元的 IEEE 754 浮點格式儲存。
3. 為什麼 `float` 可能產生誤差？因為許多十進位小數無法用有限的二進位位元精確表示，只能近似儲存。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。
Python 的 int

強調正確性優先。
盡可能精確表示整數，不因位數增加而失真。

AI 使用的 FP16、BF16

強調效率優先。
犧牲部分精度，換取更快運算速度與更少記憶體使用。


2. 各自的優點與缺點。
Python int

優點：

不易溢位。
整數運算結果精確。

缺點：

需要較多記憶體。
大數運算速度較慢。
FP16、BF16

優點：

記憶體需求小。
運算速度快。
適合大型 AI 模型。

缺點：

有精度損失。
可能產生數值誤差。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？
因為模型有數十億甚至數千億個參數，使用 FP16 或 BF16 可以大幅減少記憶體消耗並提升運算速度，而準確率通常仍能維持在可接受範圍。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？
需要更多記憶體。
訓練速度變慢。
推論速度變慢。
GPU 可容納的模型規模變小。
訓練與部署成本提高。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

Python 的 int 重視正確性，所以：

Python
x = 10**1000

仍能正確儲存超大整數，因為會自動使用更多記憶體。

AI 模型則重視效率，所以通常使用 FP32、FP16、BF16。

記憶體使用量：FP16、BF16 佔用空間較小。
計算速度：低精度浮點數運算更快。
數值精度：比高精度格式低，但通常已足夠。
GPU 效率：現代 GPU 對 FP16、BF16 有專門加速。

因此：

Python 的 int 是「正確性優先」，而 AI 模型的 FP32、FP16、BF16 是「效率優先」。
